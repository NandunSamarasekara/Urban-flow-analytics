Setup and Speed Deviation

```python
from pyspark.sql import functions as F

trips_valid = (
    spark.table("workspace.silver.trips_enriched")
    .filter(F.col("is_valid_trip"))
    .withColumn("hour_of_day", F.hour("pickup_timestamp"))
    .withColumn(
        "trip_duration_hr",
        (F.unix_timestamp("dropoff_timestamp") - F.unix_timestamp("pickup_timestamp")) / 3600.0
    )
    .filter(F.col("trip_duration_hr") > 0)
    .withColumn("speed_mph", F.col("distance_miles") / F.col("trip_duration_hr"))
)
```

```python
%sql
SELECT * FROM silver.trips_enriched
```

Hourly Speed Aggregation

```python
hourly_speed = (
    trips_valid
    .groupBy("hour_of_day")
    .agg(F.round(F.avg("speed_mph"), 2).alias("avg_speed_mph"))
)
```

Percentile based Classification

```python
thresholds = hourly_speed.select(
    F.expr("percentile_approx(avg_speed_mph, 0.25)").alias("p25"),
    F.expr("percentile_approx(avg_speed_mph, 0.75)").alias("p75")
).collect()[0]

p25, p75 = thresholds["p25"], thresholds["p75"]

dim_time_bucket = hourly_speed.withColumn(
    "bucket_label1",
    F.when(F.col("avg_speed_mph") <= p25, "Peak")
    .when(F.col("avg_speed_mph") > p25, "Off-Peak")
    .otherwise("Normal")
)
```

```python
#Saving the table
dim_time_bucket.write.format("delta").mode("overwrite").saveAsTable("workspace.gold.dim_time_bucket")
```

Creating Time Bucket Dimention: dim_time_bucket

```python
trips_valid = (
    spark.table("workspace.silver.trips_enriched")
    .filter(F.col("is_valid_trip"))
    .withColumn("hour_of_day", F.hour("pickup_timestamp"))
)

dim_time_bucket = spark.table("workspace.gold.dim_time_bucket")

trips_bucketed = trips_valid.join(
    dim_time_bucket.select("hour_of_day", "bucket_label1"),
    "hour_of_day",
    "left"
)
```

```python
od_bucket_speed = (
    trips_bucketed
    .filter(F.col("bucket_label1").isin("Peak", "Off-Peak"))
    .groupBy("pu_loc_id", "do_loc_id", "bucket_label1")
    .agg(F.avg("avg_speed_mph").alias("avg_speed"))
)
```

```python
od_pivot = (
    od_bucket_speed
    .groupBy("pu_loc_id", "do_loc_id")
    .pivot("bucket_label1", ["Peak", "Off-Peak"])
    .agg(F.first("avg_speed"))
    .withColumnRenamed("Peak", "peak_avg_speed_mph")
    .withColumnRenamed("Off-Peak", "baseline_avg_speed_mph")
)
```

```python
od_slowdown = (
    od_pivot
    .filter(F.col("peak_avg_speed_mph").isNotNull() & F.col("baseline_avg_speed_mph").isNotNull())
    .withColumn(
        "slowdown_pct",
        F.round(
            (F.col("baseline_avg_speed_mph") - F.col("peak_avg_speed_mph"))
            / F.col("baseline_avg_speed_mph") * 100, 2
        )
    )
)
```

```python
od_variability = (
    trips_valid
    .groupBy("pu_loc_id", "do_loc_id")
    .agg(
        F.stddev("trip_duration_min").alias("stddev_duration"),
        F.avg("trip_duration_min").alias("avg_duration"),
        F.count("*").alias("trip_count")
    )
    .withColumn(
        "travel_time_variability_index",
        F.round(F.col("stddev_duration") / F.col("avg_duration"), 3)
    )
    .select("pu_loc_id", "do_loc_id", "travel_time_variability_index", "trip_count")
)
```

Creating the fact table for OD congestion: fact_od_congestion

```python
fact_od_congestion = (
    od_slowdown
    .join(od_variability, ["pu_loc_id", "do_loc_id"], "inner")
    .withColumnRenamed("pu_loc_id", "origin_loc_id")
    .withColumnRenamed("do_loc_id", "dest_loc_id")
    .select(
        "origin_loc_id", "dest_loc_id",
        "baseline_avg_speed_mph", "peak_avg_speed_mph", "slowdown_pct",
        "travel_time_variability_index", "trip_count"
    )
)

fact_od_congestion.write.format("delta").mode("overwrite") \
    .saveAsTable("workspace.gold.fact_od_congestion")
```

Creating the fact table for daily run vehicles: fact_vehicle_daily

```python
EV_RANGE_KM = 300 #300 was selected because it's the least range for EVs in practical environments
```

```python
EV_RANGE_KM_TEST_1 = 200 #Assuming EVs have a range of 200km (100kms lesser than the least level)
```

```python
fact_vehicle_daily = (
    fact_table_daily.withColumn("total_daily_distance_km", F.round(F.col("total_daily_distance_miles") * 1.60934, 2))
    .withColumn("is_under_ev_range", F.col("total_daily_distance_km") <= EV_RANGE_KM) #Change this variable to test different levels
    .drop("total_daily_distance_km")
    .withColumnRenamed("pickup_date", "date")
)
```

Fact table to connect zone with hubs: fact_zone_hub_connectivity

```python
fact_zone_hub_connectivity = (
    spark.table("workspace.silver.trips_enriched")
    .filter(F.col("is_valid_trip"))
    .groupBy("pu_loc_id")
    .agg(
        F.round(F.avg(F.col("is_hub_connected").cast("int")) * 100, 2).alias("connectivity_rate_pct"),
        F.round(F.avg("pickup_nearest_hub_km"), 3).alias("avg_distance_to_hub_km"),
        F.count("*").alias("trip_count")
    )
    .withColumnRenamed("pu_loc_id", "loc_id")
)
```

```python
fact_zone_hub_connectivity.write.format("delta").mode("overwrite") \
    .saveAsTable("workspace.gold.fact_zone_hub_connectivity")
```

Checking NULLs

```python
def null_summary(table_name):
    df = spark.table(f"workspace.gold.{table_name}")
    total = df.count()
    null_counts = df.select([
        F.sum(F.col(c).isNull().cast("int")).alias(c) for c in df.columns
    ]).collect()[0].asDict()

    print(f"\n{table_name} ({total:,} rows) — null check:")
    has_nulls = False
    for col, n in null_counts.items():
        if n and n > 0:
            has_nulls = True
            print(f"{col}: {n:,} nulls ({n/total*100:.2f}%)")
    if not has_nulls:
        print("No nulls")
```

```python
for t in gold_tables:
    null_summary(t)
```

##Data Quality Check

Observation: 
* Out of 259 zones, 59 (22.87%) have a NULL value for avg_distance_to_hub_km
    - Cause: The Nearest-Hub Matching algorithm in Part A uses an H3 k-ring search (~1km radius, H3_RESOLUTION=8). For zones where no Transit Hub was found within this radius, pickup_nearest_hub_km is NULL.

* The majority of nulls are concentrated in boroughs with lower subway coverage, such as Queens (34), Staten Island (11), and the Bronx (7). This is not a data quality bug, but rather reflects a real-world disparity in city-wide transit infrastructure.

Decision: 

It was decided to keep these NULL values unchanged without widening the search radius. The fact that "no hub was found within the existing search radius for this zone" provides a meaningful signal in itself, which aligns with the decision made for Staten Island. In the Power BI dashboard, these NULL values can be explicitly labeled and displayed as "No Nearby Hub".

##Referential Integrity Check

```python
fact_od = spark.table("workspace.gold.fact_od_congestion")
dim_zone_gold = spark.table("workspace.gold.dim_zone")

orphan_origin = fact_od.join(
    dim_zone_gold, fact_od["origin_loc_id"] == dim_zone_gold["loc_id"], "left_anti"
).count()

orphan_dest = fact_od.join(
    dim_zone_gold, fact_od["dest_loc_id"] == dim_zone_gold["loc_id"], "left_anti"
).count()

print(f"fact_od_congestion — orphan origin_loc_id: {orphan_origin}")
print(f"fact_od_congestion — orphan dest_loc_id: {orphan_dest}")
```

```python
fact_vd = spark.table("workspace.gold.fact_vehicle_daily")
dim_provider_gold = spark.table("workspace.gold.dim_provider")
dim_date_gold = spark.table("workspace.gold.dim_date")

orphan_provider = fact_vd.join(
    dim_provider_gold, fact_vd["provider_code"] == dim_provider_gold["provider_code"], "left_anti"
).count()

orphan_date = fact_vd.join(
    dim_date_gold, fact_vd["date"] == dim_date_gold["date"], "left_anti"
).count()

print(f"fact_vehicle_daily — orphan provider_code: {orphan_provider}")
print(f"fact_vehicle_daily — orphan date: {orphan_date}")
```

```python
fact_hub = spark.table("workspace.gold.fact_zone_hub_connectivity")

orphan_zone = fact_hub.join(
    dim_zone_gold, fact_hub["loc_id"] == dim_zone_gold["loc_id"], "left_anti"
).count()

print(f"fact_zone_hub_connectivity — orphan loc_id: {orphan_zone}")
```
